```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([[0],
              [1],
              [1],
              [0]])
```

```python
np.random.seed(42)
input_dim = 2
hidden_dim = 2
output_dim = 1
W1 = np.random.uniform(size=(input_dim, hidden_dim))
b1 = np.zeros((1, hidden_dim))
W2 = np.random.uniform(size=(hidden_dim, output_dim))
b2 = np.zeros((1, output_dim))
W2
```

```python
print("NEuron input 1 weights: ", W1)
print("NEuron  1  2 bais: ", b1)
print("NEuron hidden 1 weights: ", W2)
print("NEuron hidden 1 weights: ", b2)
```

```python
lr = 0.5
epochs = 10
```

```python
for epoch in range(10):
    Output_Hidden_layer1 = np.dot(X, W1) 
    activation_output = sigmoid(Output_Hidden_layer1)

    Output_layer_summation = np.dot(activation_output, W2) 
    output = sigmoid(Output_layer_summation)

    # Now compute the loss

    loss = np.average((y-output)**2)
    error = output-y
    print(output)
    
    derivative_hidden_weight =sigmoid_derivative(output)*(error)
    
   

    derivative_input_weights = np.dot(derivative_hidden_weight, W2.T)
    
    derivative_input_weights = derivative_input_weights*sigmoid_derivative(activation_output)
    derivative_hidden_weights = np.dot(activation_output.T, derivative_hidden_weight)
    weights = np.dot(X.T,derivative_input_weights)
        #print(W2)
    W1 = W1- lr*weights
    W2 = W2 - lr*derivative_hidden_weights
    # if epoch % 1 == 0:
    #         print(f"Epoch {epoch}, Loss: {loss:.4f} ,w1: {W1}, w2: {W2}")
```

```python
def Prediction(x):
    z = np.dot(x,W1)+b1
    z = sigmoid(z)
    z = np.dot(z,W2)+b2
    return 1 if sigmoid(z)>0.5 else 0
```

```python
import numpy as np
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([[0],
              [1],
              [1],
              [0]])
```

```python
def sigmo(x):
    return 1/(1+np.exp(-x))
def der_sigmo(x):
    return x*(1-x)

```

```python
input_ =2
hidden_neu = 2
outp = 1
w1 = np.random.uniform(size=(input_,hidden_neu))
w2 = np.random.uniform(size=(hidden_neu,outp))


```

```python
for i in range(100):
    sum_ist = np.dot(X,w1)
    op1 = sigmo(sum_ist)

    sum_sec = np.dot(op1,w2)
    op2 = sigmo(sum_sec)


    loss = np.mean((y - op2) ** 2)
    w2_u = np.dot(op1.T, (y-op2)*der_sigmo(op2))
    w1_u = np.dot((y-op2)*der_sigmo(op2), w2.T)
    w1_u = der_sigmo(op1)* w1_u
    w1_u = np.dot( X.T,w1_u,)


    w2 -= 0.5*w2_u
    w1 -= 0.5*w1_u
    
    print(f"Epoch {i}, Loss: {loss:.4f}")
# Final output after training
```
