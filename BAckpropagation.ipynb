```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x * (1 - x)
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]])
y = np.array([[0],
              [1],
              [1],
              [0]])
np.random.seed(42)
input_dim = 2
hidden_dim = 2
output_dim = 1
W1 = np.random.uniform(size=(input_dim, hidden_dim))
b1 = np.zeros((1, hidden_dim))
W2 = np.random.uniform(size=(hidden_dim, output_dim))
b2 = np.zeros((1, output_dim))
```

```python
W2
```

```python
lr = 0.1
epochs = 100
```

```python
for epoch in range(epochs):

    # ---- Forward Pass ----
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)  # prediction

    # ---- Compute Loss (MSE) ----
    loss = np.mean((y - a2) ** 2)
    # ---- Backpropagation ----
    d_a2 = (a2 - y)  # derivative of loss wrt a2
    d_z2 = d_a2 * sigmoid_derivative(a2)

    d_a1 = np.dot(d_z2, W2.T)
    d_z1 = d_a1 * sigmoid_derivative(a1)

    d_W2 = np.dot(a1.T, d_z2) #Change in weights in 2nd layer
    d_b2 = np.sum(d_z2, axis=0, keepdims=True)#Change in bias in 2nd layer



    d_W1 = np.dot(X.T, d_z1) #Change in weights in 1st layer
    d_b1 = np.sum(d_z1, axis=0, keepdims=True) #Change in bias in 1st layer
     # ---- Update Weights ----
    W2 -= lr * d_W2
    b2 -= lr * d_b2
    W1 -= lr * d_W1
    b1 -= lr * d_b1

    # Print loss every 1000 epochs
    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
# Final output after training

```

```python
def prediction(x):
    z1 = np.dot(x, W1) + b1
    a1 = sigmoid(z1)
    
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)
    return 1 if a2>0.5 else 0
```
