```python
import tensorflow as tf
from tensorflow import keras
```

```python
from keras.datasets import boston_housing
```

```python
(x_train,y_train),(x_test,y_test) = boston_housing.load_data()
```

```python
x_train.shape
```

```python
x_train[0]
```

```python
y_train[0]
```

```python



model = keras.Sequential([
    tf.keras.layers.Dense(64, input_shape=(13,), activation = 'sigmoid'),
    tf.keras.layers.Dense(32, activation = 'tanh'),
    tf.keras.layers.Dense(16, activation = 'relu'),
    tf.keras.layers.Dense(1)
]
)
```

```python
model.compile(optimizer="adam",loss='mse', metrics=['mae'])
```

```python
mean_ = x_train.mean()
std_ = x_train.std()
```

```python
mean_, std_
```

```python
x_train = (x_train)/x_train.std()
```

```python
x_train[0]
```

```python
mean_test = x_test.mean(axis =1)
std_test = x_test.std()
```

```python
x_test = (x_test)/std_test
```

```python
y_train
```

```python
history = model.fit(x_train,y_train, epochs= 50,
                       validation_split=0.10, verbose=1,
                        batch_size= 4)
```

```python
import matplotlib.pyplot as plt
plt.Figure(figsize=(4,5))
plt.plot(history.history['loss'])
#plt.plot(history.history['val_loss'])
plt.xlabel("epochs")
plt.ylabel('loss')
plt.title('training Loss')
```

```python
import matplotlib.pyplot as plt
plt.Figure(figsize=(4,5))

plt.plot(history.history['val_loss'])
plt.xlabel("epochs")
plt.ylabel('loss')
plt.title('training Loss')
```

```python
loss , mae_loss = model.evaluate(x_test,y_test)
```
